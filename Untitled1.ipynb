```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import os
import nltk

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

from wordcloud import WordCloud

nltk.download('stopwords')
from nltk.corpus import stopwords

print("All libraries imported successfully!")
```

```python
np.random.seed(42)

# Number of properties
n = 1000

# Locations
locations = [
    "Chennai",
    "Coimbatore",
    "Madurai",
    "Trichy",
    "Bangalore",
    "Hyderabad",
    "Pune",
    "Kochi",
    "Salem",
    "Tirunelveli"
]

# Property types
property_types = [
    "Apartment",
    "Villa",
    "Independent House"
]

# Generate features
area = np.random.randint(600, 3501, n)

bedrooms = np.random.choice(
    [1, 2, 3, 4, 5],
    n,
    p=[0.10, 0.30, 0.35, 0.20, 0.05]
)

bathrooms = np.maximum(
    1,
    bedrooms + np.random.choice(
        [-1, 0, 0, 0, 1],
        n
    )
)

location_data = np.random.choice(
    locations,
    n
)

property_type = np.random.choice(
    property_types,
    n,
    p=[0.55, 0.20, 0.25]
)

age = np.random.randint(
    0,
    31,
    n
)

parking = np.random.choice(
    [0, 1, 2, 3],
    n,
    p=[0.10, 0.55, 0.25, 0.10]
)

# Location price factor
location_factor = {
    "Chennai": 1.40,
    "Bangalore": 1.50,
    "Hyderabad": 1.25,
    "Pune": 1.30,
    "Coimbatore": 1.05,
    "Kochi": 1.00,
    "Madurai": 0.85,
    "Trichy": 0.80,
    "Salem": 0.75,
    "Tirunelveli": 0.65
}

# Property type factor
property_factor = {
    "Apartment": 1.00,
    "Villa": 1.35,
    "Independent House": 1.20
}

# Calculate property price
price = (
    area
    * 5000
    * np.array([
        location_factor[x]
        for x in location_data
    ])
    * np.array([
        property_factor[x]
        for x in property_type
    ])
)

# Bedroom and bathroom adjustment
price = price * (
    1
    + (bedrooms - 2) * 0.08
    + (bathrooms - 2) * 0.04
)

# Building age adjustment
price = price * (
    1 - age * 0.008
)

# Parking value
price = price + (
    parking * 150000
)

# Random market variation
price = price * np.random.uniform(
    0.90,
    1.10,
    n
)

# Round prices
price = np.round(
    price,
    -3
)

# Create DataFrame
housing_df = pd.DataFrame({
    "area": area,
    "bedrooms": bedrooms,
    "bathrooms": bathrooms,
    "location": location_data,
    "property_type": property_type,
    "age": age,
    "parking": parking,
    "price": price.astype(int)
})

# Save dataset
housing_df.to_csv(
    "urban_housing_dataset.csv",
    index=False
)

print("Housing dataset generated successfully!")
print("Number of records:", len(housing_df))

display(housing_df.head(10))
```

```python
TARGET = "price"

X = housing_df.drop(
    columns=[TARGET]
)

y = housing_df[TARGET]

numeric_features = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X.select_dtypes(
    include="object"
).columns.tolist()

print("Numerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            "passthrough",
            numeric_features
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ]
)

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

price_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ]
)

print("Machine learning pipeline created!")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
price_pipeline.fit(
    X_train,
    y_train
)

print("Random Forest model trained successfully!")
```

```python
y_pred = price_pipeline.predict(
    X_test
)

prediction_results = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred.round(0).astype(int)
})

display(
    prediction_results.head(10)
)
```

```python
mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

r2 = r2_score(
    y_test,
    y_pred
)

print("PROPERTY PRICE PREDICTION PERFORMANCE")
print("=" * 45)

print(f"Mean Absolute Error : ₹{mae:,.2f}")
print(f"Root Mean Squared Error : ₹{rmse:,.2f}")
print(f"R² Score : {r2:.4f}")
```

```python
np.random.seed(42)

positive_reviews = [
    "The neighborhood is clean and peaceful.",
    "Excellent transportation facilities and good roads.",
    "Very safe area with friendly people.",
    "Many shops and restaurants are nearby.",
    "The area has excellent schools and hospitals.",
    "Very convenient location for families.",
    "Public transport is easily available.",
    "The neighborhood is beautiful and well maintained.",
    "Good residential area with a peaceful environment.",
    "The locality is modern and comfortable.",
    "There are many parks and recreational areas.",
    "Very good connectivity to the city.",
    "The streets are clean and well maintained.",
    "A pleasant place to live with good facilities.",
    "The area has a friendly community."
]

negative_reviews = [
    "The area is very crowded during peak hours.",
    "Traffic congestion is a major problem.",
    "The neighborhood is noisy at night.",
    "There is too much pollution in this area.",
    "Parking is difficult to find.",
    "The roads need improvement.",
    "The area is expensive compared with nearby locations.",
    "Public transportation can be difficult during peak hours.",
    "The neighborhood is not well maintained.",
    "There are frequent traffic problems.",
    "Waste management needs improvement.",
    "The area can be noisy and crowded.",
    "Some roads are damaged.",
    "There are limited parking facilities.",
    "The locality needs better infrastructure."
]

neutral_reviews = [
    "The neighborhood has both good and bad areas.",
    "The location is average for residential purposes.",
    "Facilities are available but could be improved.",
    "Transportation is available in most areas.",
    "The locality is developing gradually.",
    "Some parts are quiet while others are busy.",
    "The area has moderate traffic.",
    "Schools and shops are available nearby.",
    "The neighborhood is suitable for some families.",
    "Overall facilities are average."
]

all_reviews = []

for i in range(2000):

    location = np.random.choice(
        locations
    )

    sentiment_type = np.random.choice(
        ["Positive", "Negative", "Neutral"],
        p=[0.50, 0.25, 0.25]
    )

    if sentiment_type == "Positive":
        review = np.random.choice(
            positive_reviews
        )

    elif sentiment_type == "Negative":
        review = np.random.choice(
            negative_reviews
        )

    else:
        review = np.random.choice(
            neutral_reviews
        )

    all_reviews.append({
        "location": location,
        "review": review
    })

reviews_df = pd.DataFrame(
    all_reviews
)

reviews_df.to_csv(
    "neighborhood_reviews.csv",
    index=False
)

print("Neighborhood review dataset generated!")
print("Number of reviews:", len(reviews_df))

display(reviews_df.head(10))
```

```python
stop_words = set(
    stopwords.words("english")
)

positive_words = {
    "good",
    "great",
    "excellent",
    "amazing",
    "beautiful",
    "clean",
    "safe",
    "peaceful",
    "friendly",
    "convenient",
    "nice",
    "best",
    "comfortable",
    "wonderful",
    "pleasant",
    "modern",
    "quiet",
    "easy",
    "love"
}

negative_words = {
    "bad",
    "poor",
    "terrible",
    "worst",
    "dirty",
    "unsafe",
    "crowded",
    "noisy",
    "expensive",
    "dangerous",
    "traffic",
    "pollution",
    "awful",
    "horrible",
    "far",
    "difficult",
    "problem",
    "problems"
}


def clean_text(text):

    text = text.lower()

    text = re.sub(
        r"http\S+|www\S+",
        "",
        text
    )

    text = re.sub(
        r"[^a-zA-Z\s]",
        "",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


def calculate_sentiment(text):

    words = text.split()

    positive_count = sum(
        1
        for word in words
        if word in positive_words
    )

    negative_count = sum(
        1
        for word in words
        if word in negative_words
    )

    score = (
        positive_count
        - negative_count
    )

    if score > 0:
        sentiment = "Positive"

    elif score < 0:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    return sentiment, score


reviews_df["clean_review"] = (
    reviews_df["review"]
    .apply(clean_text)
)

sentiment_results = (
    reviews_df["clean_review"]
    .apply(calculate_sentiment)
)

reviews_df["sentiment"] = (
    sentiment_results
    .apply(lambda x: x[0])
)

reviews_df["sentiment_score"] = (
    sentiment_results
    .apply(lambda x: x[1])
)

print("Sentiment analysis completed!")

display(
    reviews_df.head(15)
)
```

```python
sentiment_counts = (
    reviews_df["sentiment"]
    .value_counts()
)

print(sentiment_counts)

plt.figure(figsize=(8, 5))

sns.countplot(
    data=reviews_df,
    x="sentiment"
)

plt.title(
    "Neighborhood Review Sentiment Distribution"
)

plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")

plt.show()
```

```python
location_sentiment = (
    reviews_df
    .groupby("location")
    .agg(
        total_reviews=("review", "count"),
        positive_reviews=(
            "sentiment",
            lambda x: (x == "Positive").sum()
        ),
        neutral_reviews=(
            "sentiment",
            lambda x: (x == "Neutral").sum()
        ),
        negative_reviews=(
            "sentiment",
            lambda x: (x == "Negative").sum()
        ),
        average_sentiment_score=(
            "sentiment_score",
            "mean"
        )
    )
    .reset_index()
)

display(
    location_sentiment
)
```

```python
average_price = (
    housing_df
    .groupby("location")["price"]
    .mean()
    .reset_index()
)

average_price.columns = [
    "location",
    "average_property_price"
]

display(
    average_price
)
```

```python
location_analysis = pd.merge(
    average_price,
    location_sentiment,
    on="location",
    how="left"
)

display(
    location_analysis
)
```

```python
def get_sentiment_description(score):

    if score > 0.30:
        return "Mostly Positive"

    elif score < -0.30:
        return "Mostly Negative"

    else:
        return "Mixed / Neutral"


location_analysis[
    "location_summary"
] = location_analysis[
    "average_sentiment_score"
].apply(
    get_sentiment_description
)

display(
    location_analysis
)
```

```python
def generate_location_report(location):

    data = reviews_df[
        reviews_df["location"] == location
    ]

    property_data = housing_df[
        housing_df["location"] == location
    ]

    average_price = (
        property_data["price"].mean()
    )

    total_reviews = len(data)

    positive = (
        data["sentiment"] == "Positive"
    ).sum()

    neutral = (
        data["sentiment"] == "Neutral"
    ).sum()

    negative = (
        data["sentiment"] == "Negative"
    ).sum()

    average_sentiment = (
        data["sentiment_score"].mean()
    )

    summary = get_sentiment_description(
        average_sentiment
    )

    positive_examples = data[
        data["sentiment"] == "Positive"
    ]["review"].head(3).tolist()

    negative_examples = data[
        data["sentiment"] == "Negative"
    ]["review"].head(3).tolist()

    report = f"""

============================================================
RESIDENTIAL LOCATION SUMMARY REPORT
============================================================

Location:
{location}

------------------------------------------------------------
PROPERTY PRICE INFORMATION
------------------------------------------------------------

Average Property Price:
₹{average_price:,.0f}

Number of Properties:
{len(property_data)}

------------------------------------------------------------
NEIGHBORHOOD REVIEW ANALYSIS
------------------------------------------------------------

Total Reviews:
{total_reviews}

Positive Reviews:
{positive}

Neutral Reviews:
{neutral}

Negative Reviews:
{negative}

Average Sentiment Score:
{average_sentiment:.2f}

Overall Review Sentiment:
{summary}

------------------------------------------------------------
POSITIVE REVIEW EXAMPLES
------------------------------------------------------------

"""

    for review in positive_examples:
        report += f"• {review}\n"

    report += """

------------------------------------------------------------
NEGATIVE REVIEW EXAMPLES
------------------------------------------------------------

"""

    for review in negative_examples:
        report += f"• {review}\n"

    report += """
============================================================
"""

    return report
```

```python
all_reports = {}

for location in locations:

    report = generate_location_report(
        location
    )

    all_reports[location] = report

    print(report)
```
